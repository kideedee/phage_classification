```python
phage_scope_dir = "E:\\master\\final_project\\data\\phage_scope\\phage"
output_file = "E:\\master\\final_project\\data\\phage_scope\\all.fasta"

# List to store all records
all_records = []
processed_count = 0
error_count = 0

print("Starting FASTA file aggregation...")

for folder in os.listdir(phage_scope_dir):
    if folder not in ["Genbank", "RefSeq"]:
        continue

    folder_path = os.path.join(phage_scope_dir, folder)
    print(f"Processing folder: {folder}")

    for file in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file)

        # Skip if not a file
        if not os.path.isfile(file_path):
            continue

        try:
            # Read the FASTA file
            record = SeqIO.read(file_path, "fasta")

            # Optionally modify the record ID to include source info
            original_id = record.id
            record.id = f"{folder}_{original_id}"
            # record.description = f"Source: {folder} | Original: {record.description}"

            # Add to our collection
            all_records.append(record)
            processed_count += 1

            if processed_count % 10000 == 0:
                print(f"Processed {processed_count} files...")

        except Exception as e:
            print(f"Error processing {file_path}: {str(e)}")
            error_count += 1
            continue

# Write all records to the output file
print(f"Writing {len(all_records)} sequences to {output_file}...")

with open(output_file, "w") as output_handle:
    SeqIO.write(all_records, output_handle, "fasta")

print(f"Aggregation complete!")
print(f"Successfully processed: {processed_count} files")
print(f"Errors encountered: {error_count} files")
print(f"Total sequences in output: {len(all_records)}")
print(f"Output saved to: {output_file}")
```
